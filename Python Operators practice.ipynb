## Python Operators

```python
# + operator add values 
print(10+3)
print(90+10)
```

```python
# + can also add values and variables
sum1 = 100+50
sum2 = sum1 +250
sum3 = sum2 + sum2
print(sum1)
print(sum2)
print(sum3)
```

# Arithmetic Operator

```python
x=15
y=4
print(x+y)  # addition
print(x-y)  # subtraction
print(x*y)  # multiplication
print(x/y)  # division
print(x%y)  # modulus
print(x**y) #Exponential
print(x//y) # flooe division
```

```python
# division always return float
x=15
y=5
a=12
b=3
print(x/y)
print(a/b)
```

```python
# floor division return integer
x=15
y=5
print(x//y)
```

# Assignment operators assign value to variable

```python
x=5
x+= 5 (x=x+5)
x-=5 (x=x-5)
```

```python
#The count variable is assigned in the if statement, and given the value 5:

numbers = [1, 2, 3, 4, 5]

if (count := len(numbers)) > 3:
    print(f"List has {count} elements")
```

## Comparison Operator

```python
x=5
y=3
print(x==y)
print(x!=y)
print(x>y)
print(x<y)
print(x<=y)
print(x>=y)
```

```python
# changing comparison operator
x = 5

print(1 < x < 10)

print(1 < x and x < 10)
```

## Logical Operator

```python
# Test if a number is greater than 0 and less than 10
x=5
print(x>0 and x<10) 
# Test if a number is less than 5 or greater than 10
print(x<5 or x>10)
# Reverse the result with not
print(not(x>3 and x<10))
```

## Identity Operators

```python
# The is operator returns True if both variables point to the same object
x = ["apple", "banana"]
y = ["apple", "banana"]
z = x

print(x is z)
print(x is y)
print(x == y)
```

```python
# The is not operator returns True if both variables do not point to the same object
x = ["apple", "banana"]
y = ["apple", "banana"]

print(x is not y)
```

```python
# Difference between is and ==
x = [1, 2, 3]
y = [1, 2, 3]

print(x == y)
print(x is y)
```

## Membership Operators

```python
# Check if "banana" is present in a list:

fruits = ["apple", "banana", "cherry"]

print("banana" in fruits)
```

```python
# Check if "pineapple" is NOT present in a list:

fruits = ["apple", "banana", "cherry"]

print("pineapple" not in fruits)
```

```python
# Also works for strings
text = "Hello World"

print("H" in text)
print("hello" in text)
print("z" not in text)
```
