```python
class Animal:
    def __init__(self, name, sound):
        self.name = name
        self.sound = sound

    def display(self):
        print("Animal:", self.name)
        print("Sound:", self.sound)


# Predefined animals
dog = Animal("dog", "bark")
cat = Animal("cat", "meow")
cow = Animal("cow", "moo")

# Dictionary to store animals
animals = {
    "dog": dog,
    "cat": cat,
    "cow": cow
}

# User input
user_animal = input("Enter animal name: ")

# Check if animal exists
if user_animal in animals:
    animals[user_animal].display()
else:
    print("This animal does not exist in the class.")
```

```python
# make calculator using oop
class calculator:
    def __init__(self,a,b):
        self.a = a
        self.b = b
    def add(self):
        return self.a + self.b
    def subtract(self):
        return self.a - self.b
    def multiply(self):
        return self.a * self.b
    def divide(self):
        if self.b==0:
            return "Mathematicla Error"
        return self.a / self.b
x = int(input("Enter a:"))
y = int(input("enter b:"))
C1 = calculator(x, y)
print("sum is" , C1.add())
print("difference is" , C1.subtract())
print("multiplication is:", C1.multiply())
print("Divide is:", C1.divide())

     
    
```
