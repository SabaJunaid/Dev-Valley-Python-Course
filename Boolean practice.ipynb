## Python Booleans Practice

```python
# python Booleans
print(10<9)
print(5==5)
print(9>2)
```

```python
# Print a message based on whether the condition is True or False
a=200
b=5
if a>b:
    print("a is greater than b")
else:
    print("a is less than b")
```

```python
a= 20
b=2
if b>a:
    print("b is greater than a")
else:
    print("b is not greater than a")
```

```python
# type casting 
# evaluate string and number
print(bool("Hello")) # non empty string is True
print(bool(15))      # non zero number is True
print(bool(""))      # Empty string is False 
print(bool(0))       # zero and empty bracket is False
```

```python
# evalute 2 variables
a= "hello"
v= 24
print(bool(a))
print(bool(v))
```

```python
# function can return boolean 
def myfun():
    return True
print(myfun())
```

```python
# Print "YES!" if the function returns True, otherwise print "NO!"
def myfun():
    return True
if myfun():
    print("Yes!")
else:
    print("No!")
```

```python
# Check if an object is an integer or not:
x = 200
print(isinstance(x, int))
print(isinstance(x, float))
print(isinstance(x, str))
```

```python
class myclass():
  def __len__(self):
    return 0

myobj = myclass()
print(bool(myobj))
```

```python
class myclass1():
    def __len__(self):
        return False
myobj1 = myclass1()
print(bool(myobj1))
```
