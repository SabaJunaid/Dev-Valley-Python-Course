## Performing function of DataFrame

```python
# DataFrame 1

df = pd.DataFrame({
    'Name': ['Ali', 'Babar', 'Ayesha', 'Daud', 'Ehsan', 'Farukh'],
    'Age': [20, 21, 22, 20, 21, 23],
    'Marks': [85, 78, 92, 74, 88, 81],
    'Subject': ['Math', 'Physics', 'Biology', 'Chemistry', 'Math', 'Physics'],
    'StudyHours': [3, 4, 5, 2, 3, 4]
})

print(df)
```

## Data information 

```python
# shape shows total number of rows and column
print(df.shape)
```

```python
# Shows information about the DataFrame like (number of rows, column names, data type of each column (int, object, float), non-null values)
print(df.info())
```

```python
# Shows statistical summary of numerical columns like count, mean, std, min, quartiles, max
print(df.describe())
```

## selecting single and multiple column

```python
# Column selection
print(df["Age"]) # selecting single column
```

```python
print(df[["Name", "Marks"]]) # selecting multiple columns 
```

## Selecting rows 

```python
# selecting row by position done by iloc
print(df.iloc[0])
print(df.iloc[1])
```

```python
# selecting rows by index done by loc()
print(df.loc[0])  # First row by index
```

```python
# slicing Rows
print(df.iloc[0:1])
```

## Accessing unique values

```python
# unique values is accessed using [df.["name"].unique()]
print(df['Age'].unique())
```

## Ununiques values

```python
# # unique() and nunique()

print(df['Name'].nunique())
```

## values count

```python
# value_counts()
print(df['Marks'].value_counts())
```
