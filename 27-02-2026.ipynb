```python
class Shape:
    def __init__(self, name, colour):
        self.__name = name # private
        self.__colour = colour # private
        # getter method
    def get_name(self):
        return self.__name
    
    def get_color(self):
        return self.__colour
    # display method
    def display_info(self): 
        print("Shape Name:", self.__name)
        print("Color:", self.__colour) # parent class
# child class rectagle
class Rectangle(Shape):
        
        def __init__(self, name, colour, length, width):
         super().__init__(name, colour)   # call parent constructor
         self.__length = length          # private
         self.__width = width            # private
# getter method
        def get_length(self):
         return self.__length
    
        def get_width(self):
         return self.__width
# display method

        def display_info(self):
         super().display_info()
         print("Length:", self.__length)
         print("Width:", self.__width)
# creat object 
r = Rectangle("Rectangle", "Blue", 10, -5)
r.display_info()
```

```python
# Parent class
class Vehicle:
    def __init__(self, name, color):
        self.__name = name      # private
        self.__color = color    # private

    # Getter methods
    def get_name(self):
        return self.__name

    def get_color(self):
        return self.__color

    # Display method
    def display_info(self):
        print("Vehicle Name:", self.__name)
        print("Color:", self.__color)


# Child class
class Car(Vehicle):
    def __init__(self, name, color, speed, fuel_type):
        super().__init__(name, color)   # call parent constructor
        self.__speed = speed             # private
        self.__fuel_type = fuel_type     # private

    # Getter methods
    def get_speed(self):
        return self.__speed

    def get_fuel_type(self):
        return self.__fuel_type

    # Display method
    def display_info(self):
        super().display_info()           # parent info
        print("Speed:", self.__speed)
        print("Fuel Type:", self.__fuel_type)


# Create object
my_car = Car("Honda Civic", "Red", 180, "Petrol")
my_car.display_info()
```

```python
# example by Sir
class Employee:
    
    def _init_(self, name, employee_id, salary):
        # Private attributes (encapsulation)
        self.__name = name              # Private (double underscore)
        self.__employee_id = employee_id # Private
        self.__salary = salary           # Private
        self._department = "General"     # Protected (single underscore)
    
    # Getter methods (encapsulation)
    def get_name(self):
        return self.__name
    
    def get_employee_id(self):
        return self.__employee_id
    
    def get_salary(self):
        return self.__salary
    
    def get_department(self):
        return self._department
    
    # Setter methods with validation (encapsulation)
    def set_salary(self, new_salary):
        if new_salary > 0:
            self.__salary = new_salary
            return True
        else:
            print("Salary must be positive!")
            return False
    
    def set_department(self, new_department):
        self._department = new_department
    
    # Public methods
    def calculate_bonus(self):
        """Calculate bonus (5% of salary)"""
        return self.__salary * 0.05
    
    def display_info(self):
        """Display employee information"""
        print(f"Name: {self.__name}")
        print(f"Employee ID: {self.__employee_id}")
        print(f"Department: {self._department}")
        print(f"Salary: ${self.__salary:,.2f}")
        print(f"Bonus: ${self.calculate_bonus():,.2f}")


# Create an employee object
emp = Employee("Alice", "E123", 75000)
emp.calculate_bonus()
emp.__name='ali'
emp.get_name()
class Manager(Employee):
    
    def _init_(self, name, employee_id, salary, team_size):
        # Call parent class constructor
        super()._init_(name, employee_id, salary)
        self.__team_size = team_size      # Private attribute
        self.set_department("Management") # Using parent's protected method
    
    # Additional methods specific to Manager
    def get_team_size(self):
        return self.__team_size
    
    def set_team_size(self, new_size):
        if new_size > 0:
            self.__team_size = new_size
            return True
        return False
    
    # # Override parent method (polymorphism)
    # def calculate_bonus(self):
    #     """Override: Managers get 10% bonus + extra per team member"""
    #     base_bonus = super().calculate_bonus()  # Call parent method
    #     team_bonus = self.__team_size * 500
    #     return base_bonus + team_bonus
    
    # Override display method
    def display_info(self):
        super().display_info()  # Call parent display
        print(f"Team Size: {self.__team_size} members")
        print(f"Management Bonus: +${self.__team_size * 500:,.2f}")
m=Manager("Bob", "M456", 90000, 5)
m.display_info()
```
