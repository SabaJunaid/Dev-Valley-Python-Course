```python
class Person:
  def __init__(self, name, age):
    self.name = name
    self.__age = age # Private property ( __ ) this prefix used to make a property or data private

p1 = Person("Emil", 25)
print(p1.name)
print(p1.__age) # Private property can not be accessed outsdie the class 
```

```python
class person:
    def __init__(self, name, age):
        self.name = name
        self.__age = age

    def get_age(self):
        return self.__age
p1 = person("Ahmad", 27)
print(p1.name)
print(p1.get_age()) # private propersity can be accessed using get_ function
```

```python
class Employee:
    def __init__(self, name, salary):
        self.__name = name
        self.__salary = salary

    def get_name(self):
        return self.__name

    def get_salary(self):
        return self.__salary

    def calculate_bonus(self, bonus_percent):
        bonus = self.__salary * bonus_percent / 100
        return self.__salary + bonus
    
# Object creation
e = Employee("Ali", 50000)

print("Name:", e.get_name())
print("Salary:", e.get_salary())
print("Salary with 10% bonus:", e.calculate_bonus(5))
```

```python
class Employee:

    def __init__(self, name, salary):
        self.__name = name
        self.__salary = salary

    def get_name(self):
        return self.__name

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):   # setter method
        self.__salary = salary

    def calculate_bonus(self):
        if self.__salary < 5000:
            print("No bonus because salary is less than 5000")
            # return self.__salary
        else:
            bonus = self.__salary * 10 / 100
            return self.__salary + bonus
e = Employee("Ali", 3000)

print("Salary:", e.get_salary())
print("Total Salary:", e.calculate_bonus())
e.set_salary(10000)

print("New Salary:", e.get_salary())
print("Total Salary with Bonus:", e.calculate_bonus())
```

```python
class Student:
  def __init__(self, name):
    self.name = name
    self.__grade = 0

  def set_grade(self, grade):
    if 0 <= grade <= 100:
      self.__grade = grade
    else:
      print("Grade must be between 0 and 100")

  def get_grade(self):
    return self.__grade

  def get_status(self):
    if self.__grade >= 60:
      return "Passed"
    else:
      return "Failed"

student = Student("SABA")
student.set_grade(85)
print(student.get_grade())
print(student.get_status())
```
