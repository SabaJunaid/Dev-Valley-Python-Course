## Practice related to Strings 

```python
# string indexing
a = "Hello, World!"
print(a[8])
```

```python
# loops in string
for x in "banana":
    print(x)
```

```python
# length of string
a = "hello world!"
print(len(a))
print(a[11])
```

```python
# in keyword to check through string
t = "the best things in life are free!"
print("free" in t)
print(len(t))
print(a[6])
```

```python
# checking string using if statement 
t = "the best things in life are free!" 
if "free" in t:
    print("yes, 'free' is present" )
    print(len(t))
    print(a[8])
 
```

```python
# check string using if NOT
txt = "The best things in life are free!"
print("expensive" not in txt)
print(len(txt))
print(txt[9])
```

```python
# if statement to check if NOT
t = "The best things in life are free!"
if "expensive" not in t:
    print('expensive' "is not present")
```

```python
# slicing
a = "hello world!"
print(a[2:5])
```

```python
# slicing from the start
b = "hello world!"
print(len(b))
print(b[3])
print(b[:5])

```

```python
# slice to the end
b = "hello world!"
print(b[0:])
print(b[6])
print(len(b))
```

```python
# negative indexing
b = "Hello World!"
print(b[-5:-2])
print(len(b))
```

```python
# string modification 
# upper case
b = "saba junaid akbar riaz"
print(b.upper())
# lower case
a = "SABA JUNAID AKBAR RIAZ"
print(a.lower())
# remove white space
a = " Hello world! "
print(a.strip())
# replace
b = "ayesha"
print(b.replace("ayesha", "saba"))
# string spliting 
A = "SABA,Junaid"
print(A.split(",")) # returns list 
```

```python
# string concatination without space
a = "SABA"
b = "Juniad"
c = a+b 
print(c)
print(len(c))
print(c.replace("SABA","Ahsan")) # replace Saba with Ahsan
print(c[3])
```

```python
# string concatination with space
a = "Ayesha"
b = "Abaidullah"
c = a+" "+b
print(c)
print(len(c))
print(c[4])
```

```python
# using f string
name="saba"
age=25
print(f"my name is {name}, i am {age}")
# placeholder {} contain variable to format them
# use {} for price variable 
price = 56
print(f"the price is {price} rupees")
# using modifier in placeholder
# display price with 2 decimals
print(f"The total price is {price:.2f} rupees")
# perform math operation within placeholder
txt = f"The price is {20*25} dollars"
print(txt)

```
