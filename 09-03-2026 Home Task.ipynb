## Data Analysis Tasks

Q11: Calculate total sales by region and find which region has the highest sales


```python
dfm.groupby('Region')['Sales'].sum()
```

Region with highest sales

```python
dfm.groupby('Region')['Sales'].sum().plot(kind='bar')
# idxmax() returns the index (label) of the maximum value
```

Q12:Find the average discount given for each product category

```python
dfm.groupby('Category')['Discount'].mean()
```

Q13: Analyze customer ratings - calculate average rating by customer type (New vs Returning)

```python
dfm.groupby('CustomerType')['customer_rating'].mean()
```

Q14: Find top 5 products by quantity sold

```python
dfm.groupby('Product')['Quantity'].sum().nlargest(5).plot(kind='bar')
# nlargest(5) returns the top n largest values from a column
```

Q15: Calculate monthly sales trend by extracting month from date column (check pandas date time modules)

```python
dfm['Date'].dt.month
dfm.groupby('month')['Sales'].sum().plot(kind='pie')
```

Q16: Analyze discount impact on sales - compare average sales with and without discount

```python
dfm.groupby(dfm['Discount'] > 0)['Sales'].mean()
```

Q17: Find the most profitable product (highest total sales amount)

```python
dfm.groupby('Product')['Sales'].sum().idxmax()
```
