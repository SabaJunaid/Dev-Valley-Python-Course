## Function in Python

```python
# variable outside of a function, and use it inside the function
x = "awesome" 
def myfunc():
    print("python is " + x)
myfunc()
```

```python
x = "awesome"
def fun2():
    print("pythom is " + x)
fun2()
```

```python
x = "awesome"

def myfunc():
  x = "fantastic"
  print("Python is " + x)

myfunc()

print("Python is " + x)
```

```python
# if you use global keyword variable belong to global scope
def myfunc():
  global x
  x = "fantastic"

myfunc()

print("Python is " + x)
```

```python
# use the global keyword if you want to change a global variable inside a function.
x = "awesome"

def myfunc():
  global x
  x = "fantastic"

myfunc()

print("Python is " + x)
```
