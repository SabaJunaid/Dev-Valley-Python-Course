## Conditional

```python
# Input a month number (1–12) and print the season it belongs to
month = int(input("Enter month number (1- 12):"))
# Determine month
if month in [12, 1, 2]:
    print("Winter")
elif month in [3, 4, 5]:
    print("Spring")
elif month in [6,7, 8]:
    print("Summer")
elif month in [9, 10, 11]:
    print("Autumn")
else:
    print("Invalid month")
```

```python
# Ask the user for a password
password = input("Enter your password: ")
# Check password strength
if len(password) < 6:
    print("Too short")
elif len(password) >= 6 and len(password) < 8:
    print("Weak")
elif len(password) >= 8 and len(password) < 12:
    print("Strong")
else:
    print("Excellent")
```

```python
# Current hour
hour = int(input("Enter the current hour (0-23): "))
# Determine Part of day
if hour >= 6 and hour < 12:
    print("Morning")
elif hour >= 12 and hour < 17:
    print("Afternoon")
elif hour >= 17 and hour < 21:
    print("Evening")
elif (hour >= 21 and hour <= 23) or (hour >= 0 and hour < 6):
    print("Night")
else:
    print("Invalid hour")
```

```python
# Check if a list of numbers contains a even, odd, negative, zero, or only positive numbers.
num = int(input("Enter your number:"))
if num > 0 and num % 2 == 0:
    print("potitive even")
elif num < 0 and num % 2 !=0:
    print("Negative odd")
else:
    print("zero")
```

```python
a = int(input("Enter side a: "))
b = int(input("Enter side b: "))
c = int(input("Enter side c: "))

if a == b == c:
    print("Square")  # if we consider all sides same
elif a == b or b == c or a == c:
    print("Rectangle")  # if two sides same
elif a + b > c and a + c > b and b + c > a:
    print("Triangle")
else:
    print("None")
```
