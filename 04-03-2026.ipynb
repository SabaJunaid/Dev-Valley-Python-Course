## Checking how many dimensions the arrays have

```python
# To find dimenshion of array
w=np.array([1,2,3,4,5]) 
print(w.ndim)
e=np.array(([3,4,5],[3,8,7]))
print(e.ndim)
```

```python
# arrayname.ndim is used to find dimensions of array
a = np.array(42)
b = np.array([1, 2, 3, 4, 5])
c = np.array([[1, 2, 3], [4, 5, 6]])
d = np.array([[[1, 2, 3], [4, 5, 6]], [[1, 2, 3], [4, 5, 6]]])

print(a.ndim)
print(b.ndim)
print(c.ndim)
print(d.ndim)
```
