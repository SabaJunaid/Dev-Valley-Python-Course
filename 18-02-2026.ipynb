## Tasks: If, Elif, Else

```python
# 1. Check if a number is positive, negative, or zero

number = float(input("Enter a number: "))

if number > 0:
    print("Positive")
elif number < 0:
    print("Negative")
else:
    print("Zero")
```

```python
# 2. Determine if a person is eligible to vote (age >= 18)
age = int(input("Enter age: "))

if age >= 18:
    print("Eligible to vote")
else:
    print("Not eligible to vote")
```

```python
# 3. Print grade based on score (A/B/C/D/F)
score = int(input("Enter score: "))

if score >= 90:
    print("Grade A")
elif score >= 80:
    print("Grade B")
elif score >= 70:
    print("Grade C")
elif score >= 60:
    print("Grade D")
else:
    print("Grade F")
```

```python
# 4. Check if a number is even or odd
num = int(input("Enter number: "))

if num % 2 == 0:
    print("Even")
else:
    print("Odd")
```

```python
# 5. Find the largest of two numbers
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

if a > b:
    print("Largest number is:", a)
else:
    print("Largest number is:", b)
```

```python
# 6. Check if temperature indicates Hot/Warm/Cold
temp = int(input("Enter temperature: "))

if temp >= 35:
    print("Hot")
elif temp >= 20:
    print("Warm")
else:
    print("Cold")
```

```python
# 8. Determine if a character is a vowel or consonant
char = input("Enter a character: ").lower()

if char in ['a', 'e', 'i', 'o', 'u']:
    print("Vowel")
else:
    print("Consonant")
```

```python
# 9. Check if a number is divisible by both 3 and 5
num = int(input("Enter number: "))

if num % 3 == 0 and num % 5 == 0:
    print("Divisible by both 3 and 5")
else:
    print("Not divisible by both 3 and 5")
```

```python
# 10. Determine ticket price based on age group
age = int(input("Enter age: "))

if age < 12:
    print("Ticket price: $5")
elif age < 60:
    print("Ticket price: $10")
else:
    print("Ticket price: $7")
```

## Tasks: Break, Continue, Pass

```python
# 1. Print numbers until 5 using break

for i in range(1, 11):
    if i == 6:
        break
    print(i)
```

```python
# 2. Skip number 3 using continue

for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

```python
# 3. Loop with pass as placeholder
for i in range(1, 10):
    if i == 3:
        pass
    print(i)
```

```python
# 4. Stop loop when a negative number appears
for i in range(5, -10, -1):
    if i == -4:
        break
    print(i)
```

```python
# 5. Skip even numbers in a loop

for i in range(1, 11):
    if i % 2 == 0:
        continue
    print(i)

```

```python
# 6. Break when sum exceeds 50
total = 0

for i in range(1, 20):
    total += i
    if total > 50:
        break

print("Final Sum:", total)
```

```python
# 7. Continue for vowels only in a string
text = input("Enter a string: ")

for char in text:
    if char.lower() not in ['a', 'e', 'i', 'o', 'u']:
        continue
    print(char)
```

```python
# 9. Skip multiples of 4
for i in range(1, 21):
    if i % 4 == 0:
        continue
    print(i)
```

```python
# 10. Use pass inside an if condition
num = int(input("Enter number: "))

if num > 0:
    pass
else:
    print("Number is not positive")
```

## Tasks: Nested Conditionals

```python
# 1. Check positive/negative then even/odd
num = int(input("Enter a number: "))

if num > 0 and num % 2 == 0:
    print("Positive Even")
elif num > 0 and num % 2 != 0:
    print("Positive Odd")
elif num < 0 and num % 2 == 0:
    print("Negative Even")
elif num < 0 and num % 2 != 0:
    print("Negative Odd")
else:
    print("Zero")
```

```python
# 2. Age check for driving and insurance category
age = int(input("Enter age: "))

if age >= 18:
    print("Eligible to drive")
    
    if age < 25:
        print("High insurance category")
    else:
        print("Normal insurance category")
        
else:
    print("Not eligible to drive")
```

```python
# 3. Score check for pass and distinction
score = int(input("Enter score: "))

if score >= 40:
    print("Pass")
    
    if score >= 75:
        print("Distinction")
        
else:
    print("Fail")
```

```python
# 4. Check number range then divisibility
num = int(input("Enter number: "))

if 1 <= num <= 100:
    print("Number is in range (1-100)")
    
    if num % 5 == 0:
        print("Divisible by 5")
    else:
        print("Not divisible by 5")
        
else:
    print("Number is out of range")
```

```python
# 5. Weather check: temp and say weather type
temp = int(input("Enter temperature: "))

if temp >= 30:
    print("Hot Weather")
    
    if temp >= 40:
        print("Extreme Heat")
        
elif temp >= 15:
    print("Moderate Weather")
    
    if temp < 20:
        print("Cool Breeze")
        
else:
    print("Cold Weather")
    
    if temp <= 0:
        print("Freezing")
```

```python
# 6. Salary check for tax slab
salary = int(input("Enter salary: "))

if salary > 0:
    
    if salary <= 50000:
        print("No Tax")
        
    elif salary <= 100000:
        print("10% Tax")
        
    else:
        print("20% Tax")
        
else:
    print("Invalid Salary")
```

```python
# 7. Check login: username then password
username = input("Enter username: ")
password = input("Enter password: ")

if username == "admin":
    
    if password == "1234":
        print("Login Successful")
    else:
        print("Wrong Password")
        
else:
    print("Wrong Username")
```

```python
# 8. Check marks then scholarship eligibility
marks = int(input("Enter marks: "))

if marks >= 50:
    print("Passed")
    
    if marks >= 85:
        print("Eligible for Scholarship")
    else:
        print("Not Eligible for Scholarship")
        
else:
    print("Failed")
```
