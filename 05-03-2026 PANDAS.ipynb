# Series Creation 

```python
import pandas as pd
# creating series with default label 
a=pd.Series([1,2,3,4])
a
```

```python
# finding index of entries in series
print(a[0])
```

```python
# creating your label
a=pd.Series([1,2,3,4], index=['x', 'y', 'z', 's'])
a
```

```python
# finding item using our label name
print(a ["x"])
```

```python
# above example can also be executed as 
a=[1,2,3,4]
myser=pd.Series(a, index=['x', 'y', 'z', 's'])
print(myser)
```

```python
# accessing item by its label name
print(myser ['y'])
```

```python
# creating series from dictionary
calories={"day1": 420, "day2": 380, "day3": 390}
myser=pd.Series(calories)
print(myser)
```

```python
# Create a Series using only data from "day1" and "day2"
calories = {"day1": 420, "day2": 380, "day3": 390}

myser=pd.Series(calories, index = ["day1", "day2"])
print(myser)
```

```python
# import data from 2 series
data={"calories": [420, 380, 390], "Duration": [50, 40, 45]}
myser=pd.DataFrame(data)
print(myser)
```

```python
# loc () is used to get one or more row from data loc() will always return row or rows
print(myser.loc[0])
```

```python
# Loc () function also used to return more than one rows
print(myser.loc[[0, 1]])
```

```python
# Step 1: Create simple data (20 rows, 10 columns)
data = {
    'Col1': [i for i in range(1, 21)],
    'Col2': [i+1 for i in range(1, 21)],
    'Col3': [i+2 for i in range(1, 21)],
    'Col4': [i+3 for i in range(1, 21)],
    'Col5': [i+4 for i in range(1, 21)],
    'Col6': [i+5 for i in range(1, 21)],
    'Col7': [i+6 for i in range(1, 21)],
    'Col8': [i+7 for i in range(1, 21)],
    'Col9': [i+8 for i in range(1, 21)],
    'Col10': [i+9 for i in range(1, 21)],
}

# Step 2: Convert to DataFrame
df = pd.DataFrame(data)

# Step 3: Add 1-2 missing values
df.loc[0, 'Col3'] = None   # first missing value
df.loc[5, 'Col7'] = None   # second missing value

# Step 4: See the DataFrame
print(df)
print(df.info())  # optional: to check missing values
```
