```python
# lambda arguments : expression
add = lambda a, b: a + b
print(add(2, 3))
square = lambda x: x * x
print(square(4))
```

```python
even = lambda n: n % 2 == 0
print(even(6))
```

```python
upper = lambda name: name.upper()
print(upper("saba"))
```

```python
def greet(name):
    print("Hello", name)
greet("Saba")
```

```python
def welcome(xyz):
    print("Hy there", xyz)
welcome("Saba")

```
