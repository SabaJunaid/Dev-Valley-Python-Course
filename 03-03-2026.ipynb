# convert 1D to 3D and perfrom indexing and slicing

```python
arr1 = np.array([1,2,3,4,5,6,7,8], [])
arr3d = arr1.reshape(2,2,2)
print(arr3d) 
```

## indexing in 3D

```python
print(arr3d[0,0,0])  # 1st block, 1st row, 1st column → 1
print(arr3d[1,1,1])  # 2nd block, 2nd row, 2nd column → 8
print(arr3d[])
```

# Slicing in 3D

```python
print(arr3d[0:1, 0:2, 0:2])
```
